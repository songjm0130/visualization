```python
# load libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy 
import PIL
import requests
```

```python
#sample data

np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(['Luffy', "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
#have two plots next to each other, so define two axes and their relative position by determine column and rows

fig,(ax1, ax2) = plt.subplots(ncols = 2, 
                             nrows = 1, 
                             figsize=(7,3))
ax1.scatter(x1, y1)
ax2.bar(x2,y2)
```

```python
plt.style.use('ggplot')
fig,(ax1, ax2) = plt.subplots(ncols = 2, 
                             nrows = 1, 
                             figsize=(7,3))
ax1.scatter(x1, y1)
ax1.set_title('Random age plot', loc='right')

ax2.bar(x2,y2)
ax2.set_title('Random saving plot', loc='right')
```

```python
#use mosaic can modify multiple axes by specifying them (ax1 is top left, ax2 is top right, ax3 is top to bottom on the left)
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'], 
                                    ['ax2', 'ax3']], 
                                    figsize = (7,4))
someaxes['ax1'].scatter(x1, y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)
plt.show()
```

```python
#Layout
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'], 
                                    ['ax2', 'ax3']], 
                                    figsize = (7,4),
                                    layout = 'constrained')
someaxes['ax1'].scatter(x1, y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)

someaxes['ax1'].set_xlabel ('A BIG LABEL', fontsize=18)
someaxes['ax2'].set_xlabel ('Another label', fontsize=18)
someaxes['ax3'].set_xlabel ('LABEL 2', fontsize=18)

#2 main layouts: tight layout, contrained layout
```

```python
#multiple vis on one axes

x = np.array(['Luffy', "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

#define fiture and axes
fig,ax = plt.subplots(figsize=(7,3))

# now call both bar and plots
ax.bar(x, y1,
       color = 'indigo')
ax.plot(x, y2,
        color = 'red')
```

```python
#add error information
y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x,y2, color ='red')

ax.errorbar (x,
             y2,
             yerr = y2_sd,
             fmt='none', 
             ecolor='indigo',
             elinewidth=4,
             capsize=6,
             capthick=4)
```

```python
#specify intervals using errorevery
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x,y2, color ='red')
ax.errorbar (x,
             y2,
             yerr = y2_sd,
             fmt='none', 
             ecolor='indigo',
             elinewidth=4,
             capsize=6,
             capthick=4, 
             errorevery=2)
```

```python
#load libraries
from PIL import Image #to get images from URL
from io import BytesIO # to store images
```

```python
response = requests.get('https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png')
image_file = BytesIO(response.content)
image = Image.open(image_file)
```

```python
fig, ax = plt.subplots(figsize = (7,3))
ax.plot(x,y2, color='red')

#overlay a new axis on top of ax as a container for the image
ax_image = fig.add_axes([0.1, #x coordinate
                         0.11, #y coordinate
                         0.15, #image width (in percent)
                         0.35]) #image height

ax_image.imshow(image)
ax_image.axis('off')
```
