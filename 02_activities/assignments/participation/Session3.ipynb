```python
# load libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy 
import PIL
import requests
```

```python
# y1 and y2 are a random integer between 0 and 100 (a total of 50)
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0, 100, 50)
y2 = np.random.randint(0, 100, 50)
```

```python
# Make a line plot with both y on same axes
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y1)
ax.plot(x,y2)
```

```python
#add a legend

fig, ax = plt.subplots(figsize = (5,3))
ax.plot (x, y1, label = "Person 1")
ax.plot (x, y2, label = "Person 2")
ax.legend (loc="lower right")
```

```python
#modify the legend
fig, ax = plt.subplots(figsize = (5,3))
ax.plot (x, y1, label = "Person 1")
ax.plot (x, y2, label = "Person 2")
ax.legend (loc="lower right",
           frameon = True, #add frame around the legend
           fontsize = 12,
           ncol = 2, #specify number of columns
           shadow = True)
```

```python
#move the legend outside the graph
fig, ax = plt.subplots(figsize = (5,3))
ax.plot (x, y1, label = "Person 1")
ax.plot (x, y2, label = "Person 2")
ax.legend (loc='upper right',
           bbox_to_anchor = (1,-0.2))
# upper right refers to the particular corner as the anchor point, (1,1) is the coordinate according to the anchor
```

```python
#Text and Annotations

#add annotation
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter (x, y1, label = "Person 1")
ax.scatter (x, y2, label = "Person 2")
ax.legend (loc='upper right',
           bbox_to_anchor = (1,-0.2))

#speficy a position (10,95) for annotation
ax.text(10, 95, "This value is important",
        ha='center', #alignment of the text
        color = 'red',
        size = 20)
```

```python
# Positioning text annotations

#make an empty graph
fig, ax = plt.subplots()
ax.axis([0,10, 0, 10])

# position anodations with reference to data, axes or figure dimensions
ax.text(1,5, ". Data: (1,5)", transform=ax.transData)
ax.text(0.5, 0.1, ". Axes: (0.5, 0.1)", transform=ax.transAxes)
ax.text(0.2, 0.2, ". Figure: (0.2., 0.2)", transform=fig.transFigure)
```

```python
#Annotating with arrows /arrowprops=/

fig, ax = plt.subplots(figsize = (5,3))
ax.scatter (x, y1, label = "Person 1")
ax.scatter (x, y2, label = "Person 2")
ax.legend (loc='upper right',
           bbox_to_anchor = (1,-0.2))

ax.annotate ('This is important!', xy=(10, 95), xytext=(20,94),
             arrowprops=dict(facecolor='black', shrink=0.05))
```

```python
#remove tick marks and labels

fig, ax = plt.subplots(figsize = (5,3))
ax.scatter (x, y1, label = "Person 1")
ax.scatter (x, y2, label = "Person 2")
ax.legend (loc='upper right',
           bbox_to_anchor = (1,-0.2))


ax.yaxis.set_major_locator(plt.NullLocator()) #remove tick and labels
#ax.xaxis.set_major_formatter(plt.NullFormatter()) #remove just labels
#ax.xaxis.set_major_locator(plt.MaxNLocator(3)) #specify maximum locators
ax.xaxis.set_major_locator(plt.MultipleLocator(5)) #specify intervals at every 5) 
```

```python
#rotate the axis labels

fig, ax = plt.subplots(figsize = (5,3))
ax.scatter (x, y1, label = "Person 1")
ax.scatter (x, y2, label = "Person 2")
ax.legend (loc='upper right',
           bbox_to_anchor = (1,-0.2))

plt.xticks(rotation=45, ha='right')
```
