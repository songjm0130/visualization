# Ontario College KPI Performance Indicators (2021-2022 Graduates) - Tablaeu Public

Here is a Tablaeu Public visualization of the graphs that compared the following four KPI categories: 
1. Graduation Rate
2. Graduation Satisfaction
3. Employment Rate at 6 Months afte Graduation
4. Employer Satisfaction
These KPIs are independently gathered and are part of the 2022-23 key performance indicators for college graduates who graduated in 2021-22 .
The data allows for side-by-side comparisons among colleges, facilitating informed decisions for prospective students, parents and stakeholders.

**Interactive features**: 
> *Highlight specific colleges*: Users can click on a college name or marker, or use the searching dropdown on the right panel to highlight its data across all performance categories. This helps in focusing on one institution while visually comparing it to others.

> *Rank colleges by KPI performance in Each Category*: Each KPI (e.g., graduation rate, employment rate) is presented in a way that allows users to see a ranked list. Users can quickly identify top- and bottom-performing colleges in each category by clicking the sort function

> *Filter function*: The dashboard provides filter sliders on the right panel to set minimum or maximum values for each metric. This lets users focus only on colleges that meet specific performance criteria (e.g., employment rate ≥ 80%).

> *Compare with provincial reference*: A provincial reference line is shown for each category. Users can instantly compare how each college performs relative to the Ontario-wide average.

> *Hover tooltips for details*: Hovering over a data point shows detailed tooltips with exact college and the KPI data.

> *Responsive layout*: The visual adjusts based on screen size and maintains interactive elements across views.
