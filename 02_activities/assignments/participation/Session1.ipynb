```python
# load libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy 
import PIL
import requests
```

```python
# y is a random integer between 0 and 100 (a total of 50)
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100, 50)
```

```python
#generate a scatterplot, width=5, length = 3
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x, y)
```

```python
#make a bar chart
fig, ax = plt.subplots(figsize = (5,3))
ax.bar(x, y)
```

```python
#horizontal bar
fig, ax = plt.subplots(figsize = (5,3))
ax.barh(x, y)
```

```python
#line graph
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y)
```

```python
# histogram: counting the frequency of the y values only, so x is not required
fig, ax = plt.subplots(figsize = (5,3))
ax.hist(y)
```

```python
#adding title and label to the graph
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y)

ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')

#make sure the labels fit in one graph
fig.tight_layout( )
```

```python
#define dictionary {} for fonts
font1 = {'family':'sans-serif','color':'blue',
         'size':20}

font2 = {'family':'monospace','color':'green',
         'size':14}
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)

#loc can set alignment:'left', 'right'. long title does not wrap automatically
ax.set_title('Total growth over time', fontdict = font1, loc='left')
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
fig.tight_layout()
```

**Modifying our data points**

```python
fig, ax = plt.subplots(figsize = (5,3))

ax.scatter(x, y,
           marker = '*',
           color = 'indigo')
```

```python
fig, ax = plt.subplots(figsize=(5,3))

#change line style and line width
#modify color with named color or hex codes
ax.plot(x,y,
        marker = '*',
        color = '#7425b9',
        linestyle = '--',
        linewidth = 2)
```

```python
fig, ax = plt.subplots(figsize=(5,3))

#marker size; marker edge color and face color
ax.plot(x,y,marker = '*',
        markersize = 12,
        color = '#7425b9',
        linestyle = '--',
        linewidth = 2,
        markeredgecolor = '#fa9359',
        markerfacecolor = '#000000')

#default grid
#ax.grid ()
ax.grid(axis = 'y', color='blue', 
        linewidth = 2, linestyle = '-.')
```

```python
# libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
 
# Create a dataframe
df = pd.DataFrame({'group':list(map(chr, range(65, 85))), 'values':np.random.uniform(size=20) })
 
# Reorder it based on values:
ordered_df = df.sort_values(by='values')
my_range=range(1,len(df.index)+1)
 
# Create a color if the group is "B"
my_color=np.where(ordered_df ['group']=='B', 'orange', 'skyblue')
my_size=np.where(ordered_df ['group']=='B', 70, 30)
 
# The horizontal plot is made using the hline() function
plt.hlines(y=my_range, xmin=0, xmax=ordered_df['values'], color=my_color, alpha=0.4)
plt.scatter(ordered_df['values'], my_range, color=my_color, s=my_size, alpha=1)
 
# Add title and axis names
plt.yticks(my_range, ordered_df['group'])
plt.title("What about the B group?", loc='left')
plt.xlabel('Value of the variable')
plt.ylabel('Group')

# show the graph
plt.show()
```
