# Ontario College Key Performance Indicators (2021-2022 Graduates)

Data source: https://data.ontario.ca/dataset/college-graduate-outcomes-survey/resource/54db4f7e-442c-41fe-aafc-ea8325ccb5e3

This data is collected to compare a college's performance in each category or compare data among Ontarian colleges for 2021 to 2022 graduates.
Below you will be able to find two dynamic graphs:

**1. Ontario College KPI Summary**

> Consolidates all KPIs into a single grouped bar chart for easier comparison across indicators and institutions.

> Average KPI indicator: A set of horizontal black lines indicates the Average KPI for each college, enabling a direct visual comparison for the overall college performance.

**2. College KPI Ranks (Interactive)**

> This line graph helps users easily track the relative performance rank of colleges, and quickly identify whether a college is performing better or worse than the provincial average for each key indicator.

> Province level indicator: Only the **Province** line is shown by default in the graph, with all other college lines set to be visible only when selected. This allows users to choose which college they want to examine, reducing clutter and improving focus.


```python
#Import the library
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import plotly.graph_objects as go
import plotly.express as px
from plotly.subplots import make_subplots
```

```python
#Import the data
df = pd.read_csv("https://files.ontario.ca/mcu_2/mcu-caat-kpi-data-2022-23-college-kpis-table-1-en-2024-02-22.csv", 
                 on_bad_lines='skip',
                 skiprows = 2)

df.columns = df.columns.str.replace(r'\s*\(2021 to 2022 Grads\)', '', regex=True) #clean up the column names
```

```python
kpi_columns = df.columns[1:5] #select the KPI columns
df['Average KPI'] = df[kpi_columns].mean(axis=1) #Calculate the KPI average for each college
province_row = df[df['College'] == "Province"]
df_filtered = df[df['College'] != "Province"]
college_name = df_filtered.columns[0]
```

## Ontario College KPI Summary (2022–2023 from 2021-2022 Graduates)

**How to read and interactive with the graph**

> Each college has grouped bars representing the different KPIs with distinctive colors (shown in the legend)

> Hover over each bar to see the actual data info

> The black dots and line showed the average KPI performance for each college

```python
# Create grouped bar chart using Plotly Express with a colorblind-accessible palette
fig = px.bar(
    df_filtered,
    x=college_name,  # Use the first column as the x-axis (college names)
    y=kpi_columns,   # Use the KPI columns as the y-values
    barmode="group",
    title="Ontario College KPI Comparison (2022–23)",
    labels={college_name: "College", "value": "Percentage", "variable": "KPI"},
    height=700,
    color_discrete_sequence=px.colors.qualitative.Safe  # Color palette for accessibility
)

# Add a line for the average KPI per college
fig.add_trace(go.Scatter(
    x=df_filtered[college_name],
    y=df_filtered['Average KPI'],
    mode='lines+markers',
    name='Average KPI per College',
    line=dict(color='black', width=2),
    marker=dict(symbol='circle', size=6)
))

fig.update_layout(
    barmode='group',
    title="Ontario College KPI Summary (2022–2023 from 2021-2022 Graduates)",
    xaxis_title="College",
    yaxis_title="Percentage",
    xaxis_tickangle=45,
    font=dict(size=12),
    height=700,
    margin=dict(l=40, r=40, t=80, b=160),
    xaxis=dict(title_font=dict(size=12))
)

fig.show()
```

## Ontario College KPI Ranks (Interactive)


**How to read and interactive with the graph**

> Each college is ranked relatively for each KPI category

> By default, the dash black line is the province level of KPI for each category.

> Click the right legend to see the college you are interested in and how they perform comparing to the province level.

```python
# Create a new dataframe to hold the rank for each college and each KPI
df_ranks = pd.DataFrame()

# Calculate ranks for each KPI
for kpi in kpi_columns:
    df_ranks = df[kpi_columns].rank(ascending=True, method='min')

# Add the college names to the rank dataframe
df_ranks[college_name] = df[college_name]

df_ranks
```

```python
#transpose the df_ranks to long format for px.graph
df_long = df_ranks.melt(id_vars=["College"], value_vars=kpi_columns, var_name="KPI", value_name="Rank")
df_long
```

```python
# Generate Rank line graph
fig = px.line(
    df_long, 
    x="KPI",  # x-axis is the KPI
    y="Rank",  # y-axis is the rank (lower values are better)
    color="College",  # Different colors for each college
    markers=True,  # Show markers for each data point
    title="KPI Ranks for Colleges (2021 to 2022 Grads) Relative to Province Level - Interactive",
    color_discrete_sequence=px.colors.qualitative.Safe, # Use the colorblind color palette
    height=800,  # Set the vertical height of the plot
    width=1500   # Set the horizontal width of the plot
)

#Set the layout
fig.update_layout(
   yaxis=dict(
        range=[26, 0],              # Set the y-axis range from 25 to 1 (reversed)
        tickmode='array',            
        tickvals=list(range(1, 26)), # Specify ticks from 1 to 25 (excluding 0 and 26)
        ticktext=list(map(str, range(1, 26))),  
        showgrid=True,              # Enable gridlines
        gridcolor='lightgrey',      # Set gridline color to light grey
        ticks="outside",            # Place ticks outside the plot
    ),
    xaxis=dict(
        title_standoff=50          # the x-axis label distance from the axis
    ),
    height=1200,
    width=1500,
    template="plotly_white",
    title_x=0.5,  # Center the title
    title_font=dict(size=18),
    margin=dict(l=100, r=100, t=100, b=100)
)

#default showing no lines, encourage the use to select the one that they want to learn about
for trace in fig.data:
    if trace.name == "Province":
        trace.line.color = "black"
        trace.line.dash = "dash"
        trace.line.width = 4
        trace.marker.size = 10
        trace.showlegend = True
        trace.name = "Province" 
    else:
        trace.visible = "legendonly"  # Default all others to hidden
    
fig.show()
```
